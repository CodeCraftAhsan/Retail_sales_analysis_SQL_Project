```python
#loc works with label and iloc only works with integer
import pandas as pd
data = pd.DataFrame({
    'Name': ['Ahasan','Hasan','Arham','Nahuan','Akib'],
    'Age': [35,34,1,5,7],
    'City': ['Helsinki','stockhome','Kuopio','Lahti','Sydney']
})
print(data)
print(data.loc[3])
print(data.iloc[3])
```

```python
data.index =['A','B','C','D','E']
print(data)
print(data.loc['C'])
data.iloc[0]
```

```python
data.loc['C','Name']
data.iloc [0,2]
print(data.loc[:,['Name','Age']])
print(data.loc[['A','C']])
print(data.iloc[[0,3]])


```

```python
#slicing loc and iloc
print(data)
print(data.loc[['A','C']])
print(data.iloc[2:3,3:4])
```

```python
# conditional filtering by loc, here () is very important to remember

print(data.loc[data['Age']<15])  # one condition

print(data.loc[(data['Age']>5) & (data['City']=='Helsinki')]) # two conditions with AND

print(data.loc[(data['City'] == 'Helsinki') | (data['Age']<7)]) # TWO conditions with OR

print(data.loc[data['Age']<10,['Name','City']])  # CONDITION WITH SELECTED COLUMN  '''df.loc[condition, columns]'''এই pattern-টা খুব ভালোভাবে মনে রাখুন।


```
